```python
def get_affilation_recursive(person_aff, affiliations):

    # if person_aff.get("affiliation") is None:
    #     person_aff.append(None)

    for affiliation in affiliations:
      #print(affiliation)
      if affiliation.get("@type", "") == "Organization":
          for id_by in affiliation.get("identifiedBy", []):
            print(id_by)
            if (
                id_by.get("@type", "") == "URI"
                and id_by.get("source", {}).get("@type", "") == "Source"
                and id_by.get("source", {}).get("code", "") == "kb.se"
            ):
                uri_value = id_by.get("value")
                person_aff.append(uri_value)
                return person_aff

      nested_affiliations = affiliation.get("hasAffiliation", [])
      if nested_affiliations is not None and len(nested_affiliations) > 0:
          person_aff = get_affilation_recursive(person_aff, nested_affiliations)

    return person_aff

```

```python
# Rekursiv funktion för att hämta konkatenera nästlad affiliering typ: forskargrupp, institution, fakultet och lärosäte.
def get_freetext_affiliations(affiliation):
    #if affiliation.get("@type", "") != "Organization":
        #return None

    nested_affils = []
    affil_name = affiliation.get("name")
    if affil_name:
        nested_affils.append(affil_name)
    affil_names_by_lang = affiliation.get("nameByLang", {})
    for affil_name_by_lang in list(affil_names_by_lang.values()):
        nested_affils.append(affil_name_by_lang)
    sub_affils = affiliation.get("hasAffiliation", [])
    for sub_affil in sub_affils:
        subsub_affils = get_freetext_affiliations(sub_affil)
        if subsub_affils:
            nested_affils.extend(subsub_affils)
    return nested_affils
```

# Min lösning

```python
# förlorar key-namn högre upp i hierarkin, exv IdentifiedBy
# plattar ut hela den nästlade strukturen, vilket betyder att det finns massor av @type/value-par som förlorar sin högre nyckel
# möjligtvis bra för överblick

def unnest_dict(obj):
  i_list = []

  for key, value in obj.items():
    if isinstance(value, str):        #str
      i_list.append({key : value})
    elif isinstance(value, dict):       #dict
      for key_1, value_1 in value.items():
        if isinstance(value_1, list):
          for i in value_1:
            u = unnest_dict(i)
            for i in u:
              i_list.append(i)
        else:
          i_list.append({key_1 : value_1})
    elif isinstance(value, list):         #list
      for i in value:
        v = unnest_dict(i) #recursive step
        for i in v:
          i_list.append(i)
    else:
      print('something went wrong with function unnest_dict')

  return i_list
```

```python
# från https://www.slingacademy.com/article/python-converting-an-sqlite-database-to-json-and-vice-versa/
def insert_from_json(json_path, db_path, table_name):
    with open(json_path, 'r') as file:
        data = json.load(file)

    conn = sqlite3.connect(db_path)
    cur = conn.cursor()

    for entry in data:
        keys = ', '.join(entry.keys())
        question_marks = ', '.join(['?' for _ in entry])
        values = tuple(entry.values())
        cur.execute(f"INSERT INTO {table_name} ({keys}) VALUES ({question_marks})", values)

    conn.commit()
    conn.close()
```
